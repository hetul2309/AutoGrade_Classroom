# Lab 1: Linear Regression

In this notebook we implement linear regression from scratch using NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data
np.random.seed(42)
X = np.random.randn(100, 1)
y = 3 * X.squeeze() + 2 + np.random.randn(100) * 0.5
print('Data shape:', X.shape, y.shape)
```

```python
# Gradient descent implementation
def train_linear_regression(X, y, lr=0.01, epochs=1000):
    w = 0.0
    b = 0.0
    n = len(y)
    losses = []
    for epoch in range(epochs):
        y_pred = w * X.squeeze() + b
        loss = np.mean((y_pred - y) ** 2)
        losses.append(loss)
        dw = (2/n) * np.sum((y_pred - y) * X.squeeze())
        db = (2/n) * np.sum(y_pred - y)
        w -= lr * dw
        b -= lr * db
    return w, b, losses

w, b, losses = train_linear_regression(X, y)
print(f'Learned: w={w:.4f}, b={b:.4f}')
```

```python
# Plot the loss curve
plt.figure(figsize=(8, 4))
plt.plot(losses)
plt.title('Training Loss (MSE)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.show()
print('Final loss:', round(losses[-1], 4))
```

```python
# Evaluate on test set
X_test = np.random.randn(20, 1)
y_test = 3 * X_test.squeeze() + 2 + np.random.randn(20) * 0.5
y_pred_test = w * X_test.squeeze() + b
test_mse = np.mean((y_pred_test - y_test) ** 2)
print(f'Test MSE: {test_mse:.4f}')
```
